```python
import pandas as pd, numpy as np

tp = pd.read_csv("train_patients.csv", header=None).values
pats = np.unique(tp[:, 276].astype(int))
print(len(pats), pats)
print("Held-out patients present:", set(pats) & {104, 113, 119, 208, 210})
print(np.unique(tp[:, 275].astype(int), return_counts=True))
```

```python
def load_xy(path, n_time=275):
    a = pd.read_csv(path, header=None).values
    X = a[:, :n_time]
    y = a[:, n_time].astype(int)
    g = a[:, n_time + 1].astype(int)   # patient ID
    return X, y, g

Xb, yb, gb = load_xy("train_beats.csv")
Xp, yp, gp = load_xy("train_patients.csv")
print(Xb.shape, Xp.shape)                    # (30992, 275) (200352, 275)
print(np.unique(yb, return_counts=True))
```

```python
held_out = {104, 113, 119, 208, 210}
a = pd.read_csv("train_patients.csv", header=None).values
mask = ~np.isin(a[:, 276].astype(int), list(held_out))
clean = a[mask]
print(clean.shape, len(np.unique(clean[:, 276])))   # expect 43 patients
print(np.unique(clean[:, 275].astype(int), return_counts=True))
```
